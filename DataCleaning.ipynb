```python
# Loading the dataset to inspect its structure and identify cleaning needs
import pandas as pd

dataset_path = 'GlobalLandTemperaturesByCity.csv'

# Load the dataset
global_temperatures = pd.read_csv(dataset_path, encoding='utf-8')

# Display the head of the dataset to understand its structure
print(global_temperatures.head())
```

```python
# Converting the 'dt' column to datetime format
# Checking for missing values in the dataset

# Convert 'dt' to datetime
global_temperatures['dt'] = pd.to_datetime(global_temperatures['dt'], errors='coerce')

# Check for missing values
missing_values = global_temperatures.isnull().sum()

# Display the missing values count
print(missing_values)
```

```python
# Dropping rows with missing values in 'AverageTemperature'
global_temperatures_cleaned = global_temperatures.dropna(subset=['AverageTemperature'])

# Check for duplicates
duplicates_count = global_temperatures_cleaned.duplicated().sum()

# Display the number of duplicates
print('Number of duplicate rows:', duplicates_count)
```

```python
# Display the head of the cleaned dataset to confirm the structure and data types
print(global_temperatures_cleaned.head())
```

```python
# Check the data types of the cleaned dataset
print(global_temperatures_cleaned.dtypes)
```

```python
# Inspecting unique values in the Latitude and Longitude columns to identify formatting issues
unique_latitudes = global_temperatures_cleaned['Latitude'].unique()
unique_longitudes = global_temperatures_cleaned['Longitude'].unique()

# Display the unique values
print('Unique Latitudes:', unique_latitudes)
print('Unique Longitudes:', unique_longitudes)
```

```python
# Convert 'Latitude' and 'Longitude' columns to strings using .loc to avoid SettingWithCopyWarning
global_temperatures_cleaned.loc[:, 'Latitude'] = global_temperatures_cleaned['Latitude'].astype(str)
global_temperatures_cleaned.loc[:, 'Longitude'] = global_temperatures_cleaned['Longitude'].astype(str)

# Filtering the dataset to find entries in Latitude and Longitude that contain non-numeric characters
problematic_latitudes = global_temperatures_cleaned[
    ~global_temperatures_cleaned['Latitude'].str.replace('N', '').str.replace('S', '-').str.strip().str.isnumeric()]
problematic_longitudes = global_temperatures_cleaned[
    ~global_temperatures_cleaned['Longitude'].str.replace('E', '').str.replace('W', '-').str.strip().str.isnumeric()]

# Display the problematic entries
print("Problematic Latitude entries:")
print(problematic_latitudes)
print("\nProblematic Longitude entries:")
print(problematic_longitudes)
```

```python
# Check for remaining missing values
print(global_temperatures_cleaned.isnull().sum())

# Handle any remaining missing values if found
global_temperatures_cleaned = global_temperatures_cleaned.dropna()  # Or use fillna() as needed
```

```python
# Checking for outliers in the 'AverageTemperature' column
print(global_temperatures_cleaned['AverageTemperature'].describe())
```

```python
global_temperatures_cleaned = global_temperatures_cleaned.reset_index(drop=True)
```

```python
print(global_temperatures_cleaned.head())
```

```python
# Shift all dates forward by a fixed number of years (e.g., 100 years)
global_temperatures_cleaned['dt'] = global_temperatures_cleaned['dt'] + pd.DateOffset(years=100)
```

```python
# Update the 'dt' column to replace the year with a recent year (e.g., all dates shifted to 2023)
global_temperatures_cleaned['dt'] = global_temperatures_cleaned['dt'].apply(lambda x: x.replace(year=2023))
```

```python
import pandas as pd

# Get the original minimum and maximum year in the dataset
original_min_year = global_temperatures_cleaned['dt'].dt.year.min()
original_max_year = global_temperatures_cleaned['dt'].dt.year.max()

# Define the new target range (1900 to 2023)
new_min_year = 1900
new_max_year = 2023

# Calculate the scaling factor for mapping the original range to the new range
original_year_range = original_max_year - original_min_year
new_year_range = new_max_year - new_min_year

# If the original range is zero, set a default mapping
if original_year_range == 0:
    # Map all dates to the new_min_year (1900)
    global_temperatures_cleaned['dt'] = global_temperatures_cleaned['dt'].apply(lambda x: x.replace(year=new_min_year))
else:
    # Function to map old dates to the new range
    def map_dates_to_new_range(date):
        # Calculate the proportional position in the original range
        proportion = (date.year - original_min_year) / original_year_range
        # Map to the new range based on the proportion
        new_year = int(new_min_year + proportion * new_year_range)
        # Update the year while keeping the same month and day
        return date.replace(year=new_year)

    # Apply the function to update the 'dt' column
    global_temperatures_cleaned['dt'] = global_temperatures_cleaned['dt'].apply(map_dates_to_new_range)

# Display the first 10 rows
print(global_temperatures_cleaned.head(10))

# Display the last 10 rows
print(global_temperatures_cleaned.tail(10))

```
